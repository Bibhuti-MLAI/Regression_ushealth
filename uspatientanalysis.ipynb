```python
# Identify potential categorical variables (including binary/integer indicators) and print their value counts
categorical_cols = ['Gender', 'education', 'currentSmoker', 'BP Meds', 'prevalentStroke', 'prevalentHyp', 'diabetes','Heart-Att']

for col in categorical_cols:
    if col in df.columns:
        print(f"--- Value Counts for '{col}' ---")
        display(df[col].value_counts(dropna=False))
        print("\n")
```

Change datatype

```python
# Convert specified columns to object dtype
cat_cols = ['education', 'currentSmoker', 'BP Meds', 'prevalentStroke', 'prevalentHyp', 'diabetes']
for col in cat_cols:
    if col in df.columns:
        df[col] = df[col].astype('object')
```

```python
df.info()
```

```python
cat = []
num = []

for i in df.columns:
  if df[i].dtype == 'object':
    cat.append(i)
  else:
    num.append(i)

print("categorical variables are :", cat)
print("numerical variables are :", num)
```

```python
df.describe().T
```

```python
df[cat].describe().T
```

```python
# Standardize column names by replacing spaces with underscores and converting to lowercase
df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_').str.lower()

# Update the 'cat' and 'num' lists with the new column names
cat = [col.replace(' ', '_').replace('-', '_').lower() for col in cat]
num = [col.replace(' ', '_').replace('-', '_').lower() for col in num]

# Verify the new columns and updated lists
print("New Column Names:", list(df.columns))
print("\nUpdated Categorical List:", cat)
print("Updated Numerical List:", num)
```

```python
# Display the updated DataFrame info to confirm
df.info()
```

Missing value treatment

```python
# Impute missing values with mode and strictly preserve the 'object' datatype
for col in cat:
    if col in df.columns:
        mode_val = df[col].mode()[0]
        # Fill missing values and explicitly cast/keep as 'object' dtype
        df[col] = df[col].fillna(mode_val).astype('object')

# Confirm imputation has been done and there are no missing values in categorical columns
print("Missing values in categorical columns:")
print(df[cat].isnull().sum())
print("\nColumn Datatypes:")
print(df[cat].dtypes)
```

```python
for col in num:
  if col in df.columns:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)
```
